## 1.7 Use reshape() to create a matrix, then flatten it back to 1D.

```python
original_array = np.arange(1, 13)
print("Original 1D Array:")
print(original_array)
print("Shape:", original_array.shape)
# Reshape the 1D array into a 3 x 4 matrix
matrix = original_array.reshape(3, 4)
print("\nReshaped Matrix:")
print(matrix)
print("Shape:", matrix.shape)
# Flatten the matrix back to 1D
flattened_array = matrix.flatten()
print("\nFlattened Array:")
print(flattened_array)
print("Shape:", flattened_array.shape)
```

# Task 2 - Vectorized Arithmetic and Linear Algebra

## 2.1  Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python
A = np.array([
    [2, 3],
    [1, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)
addition = A + B
elementwise_multiplication = A * B
# Matrix multiplication
matrix_multiplication = A @ B
print("\nMatrix Addition (A + B):")
print(addition)
print("\nElement-wise Multiplication (A * B):")
print(elementwise_multiplication)
print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)
```

## 2.2 For a square matrix, find its transpose and determinant.

```python
A_transpose = A.T
A_determinant = np.linalg.det(A)
print("Matrix A:")
print(A)
print("\nTranspose of A:")
print(A_transpose)
print("\nDeterminant of A:")
print(A_determinant)
```

## 2.3 If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python
A_inverse = np.linalg.inv(A)
print("Matrix A:")
print(A)
print("\nInverse of A:")
print(A_inverse)
# Verify A @ A_inverse
identity_matrix = A @ A_inverse
print("\nA @ A_inverse:")
print(identity_matrix)
# Check whether the result is approximately an identity matrix
verification = np.allclose(identity_matrix, np.eye(A.shape[0]))
print("\nIs A @ A_inverse approximately the identity matrix?")
print(verification)
```

```python
# checks the determinant before attempting to calculate the inverse
det_A = np.linalg.det(A)
if not np.isclose(det_A, 0):
    A_inverse = np.linalg.inv(A)
    print("Inverse of A:")
    print(A_inverse)
    result = A @ A_inverse
    print("\nA @ A_inverse:")
    print(result)
    is_identity = np.allclose(result, np.eye(A.shape[0]))
    print("\nVerification using np.allclose():", is_identity)
else:
    print("Matrix A is not invertible because its determinant is zero.")
```

## 2.4  Use vectorized NumPy operations; avoid explicit Python loops.

```python
addition = A + B
elementwise_multiplication = A * B
matrix_multiplication = A @ B
transpose_A = A.T
determinant_A = np.linalg.det(A)
inverse_A = np.linalg.inv(A)
# Verify A @ A^-1 = Identity Matrix
verification = np.allclose(
    A @ inverse_A,
    np.eye(A.shape[0])
)
print("Addition:")
print(addition)
print("\nElement-wise Multiplication:")
print(elementwise_multiplication)
print("\nMatrix Multiplication:")
print(matrix_multiplication)
print("\nTranspose of A:")
print(transpose_A)
print("\nDeterminant of A:", determinant_A)
print("\nInverse of A:")
print(inverse_A)
print("\nA @ A^-1 is Identity Matrix:", verification)
```

# Task 3 - Normal Distribution and Histogram

## 3.1 Generate at least 1,000 values from a normal distribution. State the chosen mean and standarddeviation.

```python
np.random.seed(42)
chosen_mean = 50
chosen_std = 10
normal_data = np.random.normal(
    loc=chosen_mean,
    scale=chosen_std,
    size=1000
)
print("First 10 values:")
print(normal_data[:10])
print("\nNumber of values:", normal_data.size)
print("Chosen Mean:", chosen_mean)
print("Chosen Standard Deviation:", chosen_std)
```

## 3.2  Find the sample mean and sample standard deviation. Compare them with the chosen values.

```python
sample_mean = np.mean(normal_data)
sample_std = np.std(normal_data, ddof=1)
print("Chosen Mean:", chosen_mean)
print("Sample Mean:", sample_mean)
print("\nChosen Standard Deviation:", chosen_std)
print("Sample Standard Deviation:", sample_std)
mean_difference = sample_mean - chosen_mean
std_difference = sample_std - chosen_std
print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)
```

## Observation

The calculated sample mean is close to the chosen mean of 50, and the sample standard deviation is close to the chosen standard deviation of 10. Small differences are expected because the values are randomly sampled from the normal distribution.

## 3.2 Plot a histogram with it.

plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30, edgecolor="black")
plt.title("Histogram of Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()

## Observation

The histogram shows an approximately bell-shaped distribution, which is characteristic of a normal distribution. Most of the generated values are concentrated around the chosen mean of 50, while fewer values occur toward the lower and higher ends.

# Task 5 - Filtering and Querying

## Use Boolean indexing and/or query() to answer:

## 5.1 Male passengers older than 50: how many?

```python
male_over_50 = titanic[
    (titanic["Sex"] == "male") &
    (titanic["Age"] > 50)
]
count_male_over_50 = male_over_50.shape[0]
print("Number of male passengers older than 50:", count_male_over_50)
```

## 5.2 Female first-class passengers: how many, and what percentage survived?

```python
female_first_class = titanic[
    (titanic["Sex"] == "female") &
    (titanic["Pclass"] == 1)
]
total_female_first_class = female_first_class.shape[0]
survived_female_first_class = female_first_class["Survived"].sum()
survival_percentage = (
    survived_female_first_class / total_female_first_class
) * 100
print("Number of female first-class passengers:",total_female_first_class)
print("Number who survived:",survived_female_first_class)
print("Percentage survived:",survival_percentage, "%")
```

## Observation

There were 94 female passengers travelling in first class. Among them, 91 survived, giving a survival rate of approximately 96.81%. This indicates a very high survival rate for female first-class passengers in the Titanic dataset.

## 5.3 Age 20-40, Fare above overall median, and survived: how many?

```python
overall_median_fare = titanic["Fare"].median()
print("Overall median Fare:", overall_median_fare)
filtered_passengers = titanic[
    (titanic["Age"] >= 20) &
    (titanic["Age"] <= 40) &
    (titanic["Fare"] > overall_median_fare) &
    (titanic["Survived"] == 1)
]
count_passengers = filtered_passengers.shape[0]
print("Number of passengers:", count_passengers)
```

## Observation

The overall median Fare is 14.4542. After filtering passengers aged 20–40 who paid more than the overall median Fare and survived, 130 passengers satisfy all three conditions.
