# Task 2 - Vectorized Arithmetic and Linear Algebra

## 2.1  Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python
A = np.array([
    [2, 3],
    [1, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)
addition = A + B
elementwise_multiplication = A * B
# Matrix multiplication
matrix_multiplication = A @ B
print("\nMatrix Addition (A + B):")
print(addition)
print("\nElement-wise Multiplication (A * B):")
print(elementwise_multiplication)
print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)
```

## 2.2 For a square matrix, find its transpose and determinant.

```python
A_transpose = A.T
A_determinant = np.linalg.det(A)
print("Matrix A:")
print(A)
print("\nTranspose of A:")
print(A_transpose)
print("\nDeterminant of A:")
print(A_determinant)
```

## 2.3 If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python
A_inverse = np.linalg.inv(A)
print("Matrix A:")
print(A)
print("\nInverse of A:")
print(A_inverse)
# Verify A @ A_inverse
identity_matrix = A @ A_inverse
print("\nA @ A_inverse:")
print(identity_matrix)
# Check whether the result is approximately an identity matrix
verification = np.allclose(identity_matrix, np.eye(A.shape[0]))
print("\nIs A @ A_inverse approximately the identity matrix?")
print(verification)
```

```python
# checks the determinant before attempting to calculate the inverse
det_A = np.linalg.det(A)
if not np.isclose(det_A, 0):
    A_inverse = np.linalg.inv(A)
    print("Inverse of A:")
    print(A_inverse)
    result = A @ A_inverse
    print("\nA @ A_inverse:")
    print(result)
    is_identity = np.allclose(result, np.eye(A.shape[0]))
    print("\nVerification using np.allclose():", is_identity)
else:
    print("Matrix A is not invertible because its determinant is zero.")
```

## 2.4  Use vectorized NumPy operations; avoid explicit Python loops.

```python
addition = A + B
elementwise_multiplication = A * B
matrix_multiplication = A @ B
transpose_A = A.T
determinant_A = np.linalg.det(A)
inverse_A = np.linalg.inv(A)
# Verify A @ A^-1 = Identity Matrix
verification = np.allclose(
    A @ inverse_A,
    np.eye(A.shape[0])
)
print("Addition:")
print(addition)
print("\nElement-wise Multiplication:")
print(elementwise_multiplication)
print("\nMatrix Multiplication:")
print(matrix_multiplication)
print("\nTranspose of A:")
print(transpose_A)
print("\nDeterminant of A:", determinant_A)
print("\nInverse of A:")
print(inverse_A)
print("\nA @ A^-1 is Identity Matrix:", verification)
```

# Task 3 - Normal Distribution and Histogram

## 3.1 Generate at least 1,000 values from a normal distribution. State the chosen mean and standard deviation.

```python
np.random.seed(42)
chosen_mean = 50
chosen_std = 10
normal_data = np.random.normal(
    loc=chosen_mean,
    scale=chosen_std,
    size=1000
)
print("First 10 values:")
print(normal_data[:10])
print("\nNumber of values:", normal_data.size)
print("Chosen Mean:", chosen_mean)
print("Chosen Standard Deviation:", chosen_std)
```

## 3.2  Find the sample mean and sample standard deviation. Compare them with the chosen values.

```python
sample_mean = np.mean(normal_data)
sample_std = np.std(normal_data, ddof=1)
print("Chosen Mean:", chosen_mean)
print("Sample Mean:", sample_mean)
print("\nChosen Standard Deviation:", chosen_std)
print("Sample Standard Deviation:", sample_std)
mean_difference = sample_mean - chosen_mean
std_difference = sample_std - chosen_std
print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)
```

## Observation

The calculated sample mean is close to the chosen mean of 50, and the sample standard deviation is close to the chosen standard deviation of 10. Small differences are expected because the values are randomly sampled from the normal distribution.

## 3.3 Plot a histogram with it.

```python
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30, edgecolor="black")
plt.title("Histogram of Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## Observation

The histogram shows an approximately bell-shaped distribution, which is characteristic of a normal distribution. Most of the generated values are concentrated around the chosen mean of 50, while fewer values occur toward the lower and higher ends.

## 6.4 Passenger count and survival rate by Sex-Pclass.

```python
sex_pclass_summary = titanic.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)
sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)
sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"].round(2)
)
display(sex_pclass_summary)
```

## 7.3  Try to fill NaN values by imputing using: mean, median, mode, random value

```python
titanic_original = pd.read_csv("train.csv")
print("Missing Age values:",
      titanic_original["Age"].isnull().sum())
```

```python
titanic_mean = titanic_original.copy()
mean_age = titanic_mean["Age"].mean()
titanic_mean["Age"] = titanic_mean["Age"].fillna(mean_age)
print("Mean Age:", round(mean_age, 2))
print("Missing Age after mean imputation:",
      titanic_mean["Age"].isnull().sum())
```

```python
titanic_median = titanic_original.copy()
median_age = titanic_median["Age"].median()
titanic_median["Age"] = titanic_median["Age"].fillna(median_age)
print("Median Age:", median_age)
print("Missing Age after median imputation:",
      titanic_median["Age"].isnull().sum())
```

```python
titanic_mode = titanic_original.copy()
mode_age = titanic_mode["Age"].mode()[0]
titanic_mode["Age"] = titanic_mode["Age"].fillna(mode_age)
print("Mode Age:", mode_age)
print("Missing Age after mode imputation:",
      titanic_mode["Age"].isnull().sum())
```

```python
titanic_random = titanic_original.copy()
np.random.seed(42)
age_values = titanic_random["Age"].dropna().values
missing_positions = titanic_random["Age"].isnull()
random_ages = np.random.choice(
    age_values,
    size=missing_positions.sum(),
    replace=True
)
titanic_random.loc[missing_positions, "Age"] = random_ages
print("Missing Age after random imputation:",
      titanic_random["Age"].isnull().sum())
```

## Observation

Four imputation methods were tested on the missing Age values: mean, median, mode, and random-value imputation. The mean Age was approximately 29.70, the median Age was 28.00, and the mode Age was 24.00. All four methods successfully replaced the missing Age values, resulting in 0 missing Age values after imputation. Mean and median imputation are simple deterministic methods, while random-value imputation preserves more variation in the Age values.

## 7.4 For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python
fare = titanic["Fare"]
Q1 = fare.quantile(0.25)
Q3 = fare.quantile(0.75)
print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
IQR_1_5 = 1.5 * IQR
lower_bound = Q1 - IQR_1_5
upper_bound = Q3 + IQR_1_5
print("1.5 × IQR:", IQR_1_5)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
fare_outliers = titanic[
    (titanic["Fare"] < lower_bound) |
    (titanic["Fare"] > upper_bound)
]
outlier_count = fare_outliers.shape[0]
print("Number of Fare outliers:", outlier_count)
```

## Observation

For the Fare column, Q1 is 7.9104 and Q3 is 31.00, giving an IQR of 23.0896. Using the 1.5 × IQR rule, the lower bound is -26.724 and the upper bound is 65.6344. Fare values outside these bounds are considered outliers. There are 116 Fare outliers in the standard Titanic dataset.

## 9.1 Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative relationships.

```python
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]
correlation_matrix = titanic[numerical_columns].corr()
display(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)
plt.title("Correlation Heatmap of Titanic Numerical Features")
plt.tight_layout()
plt.show()
```

```python
correlation_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack()
strongest_positive = correlation_pairs.idxmax()
positive_value = correlation_pairs.max()
strongest_negative = correlation_pairs.idxmin()
negative_value = correlation_pairs.min()
print("Strongest Positive Relationship:")
print(
    f"{strongest_positive[0]} and {strongest_positive[1]} "
    f"= {positive_value:.2f}"
)
print("\nStrongest Negative Relationship:")
print(
    f"{strongest_negative[0]} and {strongest_negative[1]} "
    f"= {negative_value:.2f}"
)
```

## Observation

The correlation heatmap shows the strength and direction of linear relationships among the numerical Titanic features. FamilySize has strong positive relationships with SibSp and Parch because it is calculated using these variables. Pclass and Fare show a negative relationship, indicating that passengers in higher-numbered classes generally paid lower fares. The correlation between Survived and most numerical variables is weaker compared with these feature-to-feature relationships.

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    titanic[titanic["Survived"] == 0]["Age"],
    titanic[titanic["Survived"] == 0]["Fare"],
    label="Did Not Survive",
    alpha=0.6
)
plt.scatter(
    titanic[titanic["Survived"] == 1]["Age"],
    titanic[titanic["Survived"] == 1]["Fare"],
    label="Survived",
    alpha=0.6
)
plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend()
plt.tight_layout()
plt.show()
```

## Observation

The scatter plot shows the relationship between Age and Fare for passengers who survived and those who did not survive. Most passengers had relatively low Fare values, while a smaller number of passengers paid very high fares. Surviving passengers are present across different age groups, but higher Fare values appear to contain a relatively larger proportion of survivors. The plot also shows that Age and Fare alone do not completely explain survival.
