```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

print("Preprocessor created")
```

```python
X_train = X.loc[train_idx]
X_test = X.loc[test_idx]

preprocessor.fit(X_train)

print("Preprocessor fitted on training data")
```

```python
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)
```

```python
y_reg_train = reg_target.loc[train_idx]
y_reg_test = reg_target.loc[test_idx]

y_clf_train = clf_target.loc[train_idx]
y_clf_test = clf_target.loc[test_idx]

print("Regression:", len(y_reg_train), len(y_reg_test))
print("Classification:", len(y_clf_train), len(y_clf_test))
```

```python
X_train_cat = pd.get_dummies(
    X_train_manual[categorical_cols_manual],
    dtype=int
)

X_test_cat = pd.get_dummies(
    X_test_manual[categorical_cols_manual],
    dtype=int
)

X_test_cat = X_test_cat.reindex(
    columns=X_train_cat.columns,
    fill_value=0
)

print("Training categorical shape:", X_train_cat.shape)
print("Testing categorical shape:", X_test_cat.shape)
print("\nEncoded columns:")
print(X_train_cat.columns.tolist())
```

```python
train_mean = X_train_manual[numerical_cols_manual].mean()
train_std = X_train_manual[numerical_cols_manual].std()

print("Training means:")
print(train_mean)

print("\nTraining standard deviations:")
print(train_std)
```

```python
X_train_num = (
    X_train_manual[numerical_cols_manual] - train_mean
) / train_std

X_test_num = (
    X_test_manual[numerical_cols_manual] - train_mean
) / train_std

print("Training numerical shape:", X_train_num.shape)
print("Testing numerical shape:", X_test_num.shape)

print("\nFirst 5 standardized training rows:")
print(X_train_num.head())
```

```python
X_train_processed_manual = pd.concat(
    [X_train_num, X_train_cat],
    axis=1
)

X_test_processed_manual = pd.concat(
    [X_test_num, X_test_cat],
    axis=1
)

print("Training shape:", X_train_processed_manual.shape)
print("Testing shape:", X_test_processed_manual.shape)

print("\nFirst 5 rows:")
print(X_train_processed_manual.head())
```

```python
y_train_manual = y_reg_train.to_numpy()
y_test_manual = y_reg_test.to_numpy()

print("Training target shape:", y_train_manual.shape)
print("Testing target shape:", y_test_manual.shape)

print("\nFirst 10 training targets:")
print(y_train_manual[:10])
```

```python
X_train_np = X_train_processed_manual.to_numpy()
X_test_np = X_test_processed_manual.to_numpy()

print("Training X shape:", X_train_np.shape)
print("Testing X shape:", X_test_np.shape)
```

```python
start_time = time.perf_counter()

beta = np.linalg.inv(
    X_train_bias.T @ X_train_bias
) @ X_train_bias.T @ y_train_manual

train_time_manual_reg = time.perf_counter() - start_time

print("Number of coefficients:", len(beta))
print("Training time:", train_time_manual_reg, "seconds")
print("\nFirst 10 coefficients:")
print(beta[:10])
```

```python
start_time = time.perf_counter()

y_manual_pred = X_test_bias @ beta

predict_time_manual_reg = time.perf_counter() - start_time

print("Prediction time:", predict_time_manual_reg, "seconds")

print("\nFirst 10 predictions:")
print(y_manual_pred[:10])
```

```python
errors = y_test_manual - y_manual_pred

mae_manual_reg = np.mean(np.abs(errors))
rmse_manual_reg = np.sqrt(np.mean(errors ** 2))

ss_total = np.sum(
    (y_test_manual - np.mean(y_test_manual)) ** 2
)

ss_residual = np.sum(
    errors ** 2
)

r2_manual_reg = 1 - (ss_residual / ss_total)

print("MAE :", mae_manual_reg)
print("RMSE:", rmse_manual_reg)
print("R2  :", r2_manual_reg)
```

```python
y_clf_train_manual = y_clf_train.to_numpy()
y_clf_test_manual = y_clf_test.to_numpy()

print("Training target shape:", y_clf_train_manual.shape)
print("Testing target shape:", y_clf_test_manual.shape)

print("\nFirst 10 training targets:")
print(y_clf_train_manual[:10])
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

test_values = np.array([-2, -1, 0, 1, 2])

print(sigmoid(test_values))
```

```python
weights = np.zeros(X_train_bias.shape[1])

print("Number of weights:", len(weights))
print("First 10 weights:", weights[:10])
```

```python
learning_rate = 0.01
iterations = 1000

start_time = time.perf_counter()

for i in range(iterations):
    z = X_train_bias @ weights
    probabilities = sigmoid(z)

    gradient = (
        X_train_bias.T @ (probabilities - y_clf_train_manual)
    ) / len(y_clf_train_manual)

    weights = weights - learning_rate * gradient

train_time_manual_clf = time.perf_counter() - start_time

print("Training time:", train_time_manual_clf, "seconds")
print("\nFirst 10 weights:")
print(weights[:10])
```

```python
start_time = time.perf_counter()

z_test = X_test_bias @ weights
y_clf_prob = sigmoid(z_test)

predict_time_manual_clf = time.perf_counter() - start_time

print("Prediction time:", predict_time_manual_clf, "seconds")

print("\nFirst 10 probabilities:")
print(y_clf_prob[:10])
```

```python
y_clf_pred_manual = (y_clf_prob >= 0.5).astype(int)

print("First 20 predictions:")
print(y_clf_pred_manual[:20])

print("\nPrediction counts:")
print(pd.Series(y_clf_pred_manual).value_counts())
```

```python
tp = np.sum((y_clf_test_manual == 1) & (y_clf_pred_manual == 1))
tn = np.sum((y_clf_test_manual == 0) & (y_clf_pred_manual == 0))
fp = np.sum((y_clf_test_manual == 0) & (y_clf_pred_manual == 1))
fn = np.sum((y_clf_test_manual == 1) & (y_clf_pred_manual == 0))

accuracy_manual_clf = (tp + tn) / len(y_clf_test_manual)

precision_manual_clf = tp / (tp + fp) if (tp + fp) > 0 else 0
recall_manual_clf = tp / (tp + fn) if (tp + fn) > 0 else 0

f1_manual_clf = (
    2 * precision_manual_clf * recall_manual_clf
    / (precision_manual_clf + recall_manual_clf)
    if (precision_manual_clf + recall_manual_clf) > 0 else 0
)

print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)

print("\nAccuracy :", accuracy_manual_clf)
print("Precision:", precision_manual_clf)
print("Recall   :", recall_manual_clf)
print("F1 Score :", f1_manual_clf)
```
